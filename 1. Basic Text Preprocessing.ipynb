##4.Deal with Punctuations

```python
import string
len(string.punctuation)
```

```python
string.punctuation
```

```python
exculsion_criteria = string.punctuation
```

```python
def remove_punctuation(text):
  for char in exculsion_criteria:
    text = text.replace(char, '')
  return text
```

```python
sample_text = "one of the other reviewers has mentioned that after watching just 1 oz episode you'll be hooked. they are right, as this is exactly what happened with me.the first thing that struck me about oz was its brutality and unflinching scenes of violence, which set in right from the word go. trust me, this is not a show for the faint hearted or timid. this show pulls no punches with regards to drugs, sex or violence. its is hardcore, in the classic use of the word.it is called oz as that is the nickname given to the oswald maximum security state penitentary. it focuses mainly on emerald city, an experimental section of the prison where all the cells have glass fronts and face inwards, so privacy is not high on the agenda. em city is home to many..aryans, muslims, gangstas, latinos, christians, italians, irish and more....so scuffles, death stares, dodgy dealings and shady agreements are never far away.i would say the main appeal of the show is due to the fact that it goes where other shows wouldn't dare. forget pretty pictures painted for mainstream audiences, forget charm, forget romance...oz doesn't mess around. the first episode i ever saw struck me as so nasty it was surreal, i couldn't say i was ready for it, but as i watched more, i developed a taste for oz, and got accustomed to the high levels of graphic violence. not just violence, but injustice (crooked guards who'll be sold out for a nickel, inmates who'll kill on order and get away with it, well mannered, middle class inmates being turned into prison bitches due to their lack of street skills or prison experience) watching oz, you may become comfortable with what is uncomfortable viewing....thats if you can get in touch with your darker side."
cleaned_text = remove_punctuation(sample_text)
print(cleaned_text)
```

```python
# first 2 arguments of maketrans function is to specify mapping -- its more optimized technique
def remove_punctuation_2(sample_text):
  return sample_text.translate(str.maketrans('', '', exculsion_criteria))
```

##5.Dealing with Abbreviations/Short forms

```python
import pandas as pd
# we must use raw data from github source
url = 'https://raw.githubusercontent.com/rishabhverma17/sms_slang_translator/refs/heads/master/slang.txt'
Abbreviations = pd.read_csv(url, sep='=',header=None, names=['Abbreviation','Detail'])
Abbreviations
```

```python
abbv_dict = dict(zip(Abbreviations['Abbreviation'], Abbreviations['Detail']))
abbv_dict
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in abbv_dict:
            new_text.append(abbv_dict[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

```python
chat_conversion('HBD he is the best')
```

```python
movie_reviews['review'] = movie_reviews['review'].apply(chat_conversion)
```

```python
import string
```

```python
def chat_conversion(text):
    new_text = []
    for word in text.split():
        # Separate leading/trailing punctuation from the core word
        leading_punct = ''
        while word and word[0] in string.punctuation:
            leading_punct += word[0]
            word = word[1:]

        trailing_punct = ''
        while word and word[-1] in string.punctuation:
            trailing_punct = word[-1] + trailing_punct
            word = word[:-1]

        # Check if the cleaned word (in uppercase) is an abbreviation
        if word.upper() in abbv_dict:
            new_text.append(leading_punct + abbv_dict[word.upper()] + trailing_punct)
        else:
            new_text.append(leading_punct + word + trailing_punct)
    return " ".join(new_text)
```

### Applying TextBlob Correction to Individual Reviews

We need to define a function to apply `TextBlob.correct()` to each review. Please be aware that running this on all 50,000 reviews will be very slow.

```python
from textblob import TextBlob

def correct_single_review_textblob(text):
    try:
        return str(TextBlob(text).correct())
    except Exception as e:
        print(f"Error correcting text: {text[:50]}... Error: {e}")
        return text # Return original text if correction fails

# Let's test it on a few reviews to see the effect
print("Original review 0:", movie_reviews['review'][0])
print("Corrected review 0:", correct_single_review_textblob(movie_reviews['review'][0]))

print("\nOriginal review 1:", movie_reviews['review'][1])
print("Corrected review 1:", correct_single_review_textblob(movie_reviews['review'][1]))
```

As you can see, `TextBlob` attempts to correct each word.
```python
def remove_stopwards(text):
  new_text = []
  for word in text.split():
    if word in eng_stopwords:
      new_text.append('')
    else:
      new_text.append(word)
  return " ".join(new_text)
```

### Sentence Tokenization using Regular Expressions

```python
# The regex below splits the text at a period, question mark, or exclamation mark,
# only if that punctuation is followed by a space. The `(?<=[.!?])` is a positive lookbehind
# assertion, meaning the split occurs after the punctuation but doesn't include the punctuation
# in the split delimiter itself.

sentences_re = re.split(r'(?<=[.!?])\s+', review_1st)
for i, sentence in enumerate(sentences_re):
    print(f"Sentence {i+1}: {sentence}")
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
wnl = WordNetLemmatizer()
```

```python
movie_reviews['review'][0]
```

```python
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(review_1st)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wnl.lemmatize(word,pos='a')))
```

```python
def lemmatize_words(text):
  new_text = []
  for word in text.split():
    new_text.append(wnl.lemmatize(word,pos='v'))
  return " ".join(new_text)

lemmatize_words(review_1st)
```
