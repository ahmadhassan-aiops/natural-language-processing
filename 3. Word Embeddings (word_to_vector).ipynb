#Word Embeddings

-  In natural language processing, word embedding is a term used for the representation of words for text analysis, typically in the form of a real-valued vector **that encodes the meaning of the word** such that the words that are closer in the vector space are expected to be similar in meaning.

**Cosine similarity scores**

- They range between **-1 and 1:**

      1.0 → identical meaning (or extremely close)
      0.7 to 0.8 → very similar (like your results)
      0 → unrelated
      -1 → opposite meaning (rare in word embeddings)

```python
import re
import numpy as np

# Preprocess (clean + tokenize)
processed_docs = [
    re.findall(r'\b\w+\b', doc.lower())
    for doc in docs
]

# Convert each document to vector
doc_vectors = []

for doc in processed_docs:
    word_vecs = []

    for word in doc:
        # ignore OOV words
        if word in model:
            word_vecs.append(model[word])

    if len(word_vecs) > 0:
        doc_vec = np.mean(word_vecs, axis=0)
    else:
        doc_vec = np.zeros(300)

    doc_vectors.append(doc_vec)

# Convert to array
doc_vectors = np.array(doc_vectors)

# Output shape is 10, 300
print("Shape:", doc_vectors.shape)
print(doc_vectors[0])
```
