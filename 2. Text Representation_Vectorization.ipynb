#1. Text Representation/Vectorization Introduction

##1.1.What is feature extraction from text?

Feature extraction from text, also known as text representation or text vectorization, is the process of converting raw text data into numerical features that can be used by machine learning algorithms.


##1.2.Why is it important?
Machine learning models cannot directly process raw text. They require numerical input. Feature extraction bridges this gap by transforming human-readable text into a machine-understandable format.

##1.3.Why is this difficult to do?

Feature extraction from text is hard because human language is complex and often unclear. Unlike numbers, text does not have a direct mathematical form, so computers struggle to understand meaning. For example, a word like “bank” can mean a financial institution or the side of a river, but simple methods treat it the same. Language also includes context, sarcasm, slang, and typos, which are difficult for models to interpret correctly. On top of that, there is a huge and constantly changing vocabulary, making it harder to handle rare words and new expressions while also processing large amounts of text efficiently.

In comparison, feature extraction is easier for images and voice data because they are already in structured numerical formats. Images are made of pixel values, and voice signals are continuous waveforms, both of which can be directly processed using mathematical operations. This makes it simpler for models to extract patterns, while text requires converting unstructured language into meaningful numerical representations first, which is the main challenge.


##1.4. What is the core idea behind this?

The core idea behind text representation is simple: convert human language into numbers in a way that **keeps its meaning (semantics) intact**, this is the main goal (your “north star”). Instead of just counting words, we want models to understand how words, phrases, and sentences relate to each other. For example, “king” should be closer to “queen” than to “apple” because their meanings are related. So, the aim is to represent text such that similar meanings are close together in numerical form, allowing machines to recognize context, relationships, and intent, and ultimately perform tasks like translation, summarization, and sentiment analysis more like humans do.


##1.5. Types of Text Representation/Vectorization

Here are some important types of text representation:

*   **Bag-of-Words (BoW)**
*   **OHE**
*   **TF-IDF (Term Frequency-Inverse Document Frequency)**
*   **N-grams**
*   **Custom features**
*   **Word Embeddings** (e.g., Word2Vec, GloVe, FastText)
*   **Contextual Embeddings** (e.g., BERT, GPT, ELMo)

##2.4.TF-IDF (Term Frequency-Inverse Document Frequency)

- When we have to use this techq?

    `Lets say if one word is present more in 1 doc compare to other docs (corpus), we should be given more weightage to this word`

###What does high and low tf idf value means for a word?


- **High TF-IDF value** for a word means that the word is very relevant and significant to a specific document within the corpus. It implies the word appears frequently in that particular document (high Term Frequency) but is relatively rare across other documents in the corpus (high Inverse Document Frequency). This combination makes the word a good distinguishing feature for that document.

- **Low TF-IDF value** for a word means that the word is either not very frequent in the specific document.
Very common across many documents in the corpus (e.g., stop words like 'the', 'is', 'a'), even if it appears frequently in the current document. Words with low TF-IDF values are generally not good discriminators for a document, as they don't help differentiate it from others.

###Why use log to calculate IDF term?

- Imagine you're ranking how "special" a word is by how rare it is. Without the log, a word appearing in 1 out of 10,000 documents would be judged 100 times more important than a word appearing in 100 documents — but intuitively it's not *100 times* more meaningful, just somewhat more. Rare words are so much rarer than common ones that this raw ratio explodes, and a single super-rare word (often just a typo or a name) would completely drown out everything else in the score. The log squashes that runaway growth into gentle, sensible steps — so rarity still boosts a word's weight, but with diminishing returns instead of an exploding one, letting the term-frequency part of the formula actually still matter.

###Topic modeling using TFIDF Technique

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF

# 1. Clean HTML tags out of the reviews
clean_docs = [re.sub(r'<[^>]+>', ' ', d) for d in docs]

# 2. Vectorize
tf_idf = TfidfVectorizer(
    stop_words='english',
    max_df=0.8,                      # drop words in >80% of docs
    min_df=1,
    token_pattern=r'\b[a-z]{3,}\b',  # only alphabetic tokens, 3+ chars
    lowercase=True
)
X = tf_idf.fit_transform(clean_docs)
vocab = tf_idf.get_feature_names_out()

# 3. Topic model
n_topics = 3
nmf = NMF(
    n_components=n_topics,
    random_state=42,
    init='nndsvd',
    max_iter=1000
)
W = nmf.fit_transform(X)   # (n_docs, n_topics) — topic weights per review
H = nmf.components_        # (n_topics, n_words) — word weights per topic

# 4. Top words per topic
for i, topic in enumerate(H):
    top_words = [vocab[j] for j in topic.argsort()[::-1][:8]]
    print(f"Topic {i}: {', '.join(top_words)}")

print()

# 5. Dominant topic per review
for i, row in enumerate(W):
    print(f"Review {i}: Topic {row.argmax()} (score {row.max():.2f})")
```
